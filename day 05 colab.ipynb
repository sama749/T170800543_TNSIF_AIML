## 🔹 What is Data Cleaning?

Data Cleaning is the process of:
- Fixing incorrect data
- Handling missing values
- Removing duplicates
- Standardizing formats

---

## 🔹 Why is Data Preprocessing Important?

Machine Learning models:
- Cannot handle missing or messy data properly
- Require numerical and consistent input

---

Fix text inconsistencies

```python
df["Gender"] = df["Gender"].str.strip().str.title()
df["City"] = df["City"].str.strip().str.title()
```

```python
#Fill numerical missing values with mean

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["CGPA"] = df["CGPA"].fillna(df["CGPA"].mean())
df["Programming"] = df["Programming"].fillna(df["Programming"].mean())
df["Communication"] = df["Communication"].fillna(df["Communication"].mean())
```

```python
df.head()
```

Fill categorical missing values with mode

```python
#Fill categorical missing values with mode

df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])
df["City"] = df["City"].fillna(df["City"].mode()[0])
```

```python
df.isnull().sum()
```

Remove duplicate rows

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.shape
```

Handle outliers

```python
Q1 = df["CGPA"].quantile(0.25)
Q3 = df["CGPA"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower limit:", lower)
print("Upper limit:", upper)

df[(df["CGPA"] < lower) | (df["CGPA"] > upper)]
```

```python
def find_outliers(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    return df[(df[column] < lower) | (df[column] > upper)]

    find_outliers("Maths")
    find_outliers("Programming")
```

Handling outliers

```python
df["Maths"] = df["Maths"].clip(0, 100)
df["Programming"] = df["Programming"].clip(0, 100)
df["Communication"] = df["Communication"].clip(0, 100)

df["CGPA"] = df["CGPA"].clip(0, 10)
```

```python
#Handling Age

df.loc[(df["Age"] < 17) | (df["Age"] > 30), "Age"] = np.nan
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

Feature Engineering

```python
#Average marks

df["Average_Marks"] = df[["Maths", "Programming", "Communication"]].mean(axis=1)
```

```python
#Performance category

df["Performance"] = pd.cut(
    df["Average_Marks"],
    bins=[0, 50, 70, 85, 100],
    labels=["Poor", "Average", "Good", "Excellent"]
)
```

```python
#pass/fail

df["Result"] = np.where(df["Average_Marks"] >= 40, "Pass", "Fail")
```

```python
df.head()
```

Encoding categorical data

```python
#Label encoding for Gender

df["Gender_Encoded"] = df["Gender"].map({
    "Male": 0,
    "Female": 1
})
```

```python
#One-hot encoding for City

df = pd.get_dummies(df, columns=["City"], dtype=int)
```

```python
df.head()
```

Rename columns to lowercase

```python
df.columns = df.columns.str.lower()
df.columns
```

Convert float to integer where applicable

```python
df["age"] = df["age"].round().astype(int)
df["maths"] = df["maths"].round().astype(int)
df["programming"] = df["programming"].round().astype(int)
df["communication"] = df["communication"].round().astype(int)
```
